```python
import pandas as pd 
import matplotlib.pyplot as plt
import os
import numpy as np

YEAR=2023  # parameter, default 2023
IMSIZE=64  # parameter, default 64

main_folder = "sirta_%04d_image%d-localtime_ICHP-daytime-dhi" % (YEAR, IMSIZE)

sirta_data = pd.read_csv(os.path.join(main_folder, "sirta_data_%04d.csv" % YEAR), index_col=0)
#sirta_data["Datetime"] = pd.to_datetime(sirta_data["Datetime"])  
num_rows_bf = len(sirta_data)  
sirta_data  = sirta_data.drop_duplicates(subset=['Datetime'], keep="first")  

num_rows_af = len(sirta_data) 

print("%d [before] >> drop duplicates >> %d [now]" % (num_rows_bf, num_rows_af))

sirta_data.to_csv(os.path.join(main_folder, "sirta_data_%04d.csv" % YEAR))
```
